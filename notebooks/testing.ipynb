```python
import numpy as np
import joblib

# Load everything
binary_model = joblib.load("binary_failure_model.pkl")
failure_type_model = joblib.load("failure_type_model.pkl")
le_failure = joblib.load("failure_type_encoder.pkl")
le_type = joblib.load("type_encoder.pkl")
BEST_THRESHOLD = joblib.load("best_threshold.pkl")

def predict_machine(type_value,
                    air_temp,
                    process_temp,
                    rotational_speed,
                    torque,
                    tool_wear):

    # Encode type
    type_encoded = le_type.transform([type_value])[0]

    # Create input array
    sample = np.array([[type_encoded,
                        air_temp,
                        process_temp,
                        rotational_speed,
                        torque,
                        tool_wear]])

    # Stage 1
    prob = binary_model.predict_proba(sample)[:,1][0]

    if prob > BEST_THRESHOLD:
        failure_type_encoded = failure_type_model.predict(sample)
        failure_type = le_failure.inverse_transform(failure_type_encoded)[0]

        return {
            "Failure": True,
            "Failure Probability": float(prob),
            "Failure Type": failure_type
        }

    else:
        return {
            "Failure": False,
            "Failure Probability": float(prob),
            "Failure Type": None
        }
```

```python
result = predict_machine(
    type_value="H",
    air_temp=305,
    process_temp=325,
    rotational_speed=1600,
    torque=65,
    tool_wear=220
)

print(result)
```
